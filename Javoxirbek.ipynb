```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

# Create a second-degree polynomial regression model
mymodel = np.poly1d(np.polyfit(x, y, 2))

# Extract coefficients
a, b, c = mymodel.coefficients

# Prediction function
def f(x_value):
    return mymodel(x_value)

# Print prediction equation
print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

# Predict speed at 5:00 PM (17:00)
prediction_time = 17
predicted_speed = f(prediction_time)

# Print predicted speed
print(f"Predicted speed at 5:00 PM: {predicted_speed:.2f} km/h")

# Generate x-values for regression curve
myline = np.linspace(1, 22, 100)

# Visualization
plt.scatter(x, y, label="Actual data")
plt.plot(myline, f(myline), color='red', label="Polynomial regression")

# Display predicted value on graph
plt.scatter(prediction_time, predicted_speed, color='green', s=100,
            label=f"Prediction at 17:00 = {predicted_speed:.2f} km/h")

plt.annotate(
    f"17:00 → {predicted_speed:.2f} km/h",
    (prediction_time, predicted_speed),
    xytext=(10, 10),
    textcoords="offset points"
)

plt.title("Polynomial Regression (degree=2)")
plt.xlabel("Time Davranov Javoxirbek")
plt.ylabel("Vehicle Speed (km/h)")
plt.legend()
plt.grid(True)
plt.show()
```
